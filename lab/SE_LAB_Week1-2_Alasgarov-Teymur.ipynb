### 10. Attendance Counter


````python
attendance = [True, True, False, True, False, True, True]

present_days = 0
absent_days = 0

for day in attendance:
    if day:
        present_days += 1
    else:
        absent_days += 1

print(f"Present: {present_days}, Absent: {absent_days}")

if present_days >= 5:
    print("Good attendance")
else:
    print("Attendance warning")
````

### 11. Number Classifier

Given:

```python

```

Loop through the list.

For every value print one of:

```text

-3 is negative
0 is zero
```

At the end, print the total number of positive, negative, and zero values.


````python
numbers = [5, -3, 0, 12, -8, 7, 0, -1]

pos_count = 0
neg_count = 0
zero_count = 0

for num in numbers:
    if num > 0:
        print(f"{num} is positive")
        pos_count += 1
    elif num < 0:
        print(f"{num} is negative")
        neg_count += 1
    else:
        print(f"{num} is zero")
        zero_count += 1

print(f"Positive = {pos_count}, Negative = {neg_count}, Zero = {zero_count}")
````

### 15. Cinema Ticket Price

A cinema determines ticket price from age.

Rules:

- under `6` → free;
- `6–17` → 5 AZN;
- `18–64` → 10 AZN;
- `65+` → 6 AZN.

Given:

```python
age = 17
```

print the correct ticket price.

Then test:

```text
5
6
17
18
64
65
80
```

#### Extra condition

If the person is a student:

```python
is_student = True
```

and their normal ticket is 10 AZN, reduce it to 8 AZN.

Combine numerical conditions with the Boolean condition.


````python
ages = [5, 6, 17, 18, 64, 65, 80]
is_student = True

for age in ages:
    if age < 6:
        price = 0
    elif 6 <= age <= 17:
        price = 5
    elif 18 <= age <= 64:
        price = 10
        if is_student:
            price = 8
    else:
        price = 6
    
    print(f"Age: {age}, Price: {price} AZN")
````

### 16. ATM PIN — Three Attempts

The correct PIN is:

```python
correct_pin = "2468"
```

Give the user at most **3 attempts**.

Ask using:

```python
input()
```

If correct:

```text
Access granted
```

and stop the loop.

If all three attempts fail:

```text
Card blocked
```

The PIN must stay a **string**.


````python
correct_pin = "2468"
attempts = 3
granted = False

for i in range(attempts):
    user_input = input(f"Enter PIN: ")
    if user_input == correct_pin:
        print("Access granted")
        granted = True
        break
    else:
        print("Incorrect PIN.")

if not granted:
    print("Card blocked")
````

### 17. Bus Passenger Counter

A bus has:

```python
capacity = 40
passengers = 0
```

Passenger groups arrive:

```python
groups = [8, 12, 7, 15, 4]
```

Loop through the groups.

If adding a group does **not** exceed capacity:

```text
Group of 8 entered
```

and update the passenger count.

Otherwise:

```text
Group of 15 cannot enter
```

At the end print:

```text
Passengers: ...
Free seats: ...
```


````python
capacity = 40
passengers = 0
groups = [8, 12, 7, 15, 4]

for group in groups:
    if passengers + group <= capacity:
        passengers += group
        print(f"Group of {group} entered")
    else:
        print(f"Group of {group} cannot enter")

free_seats = capacity - passengers
print(f"Passengers: {passengers}")
print(f"Free seats: {free_seats}")
````

### 18. Palindrome Checker

A palindrome reads the same forwards and backwards.

Examples:

```text
level
radar
madam
```

Given:

```python
word = "level"
```

print:

```text
Palindrome
```

or:

```text
Not palindrome
```

Slicing is allowed here.

Test:

```text
level
python
radar
hello
madam
```

#### Bonus

Test:

```python
text = "Never Odd Or Even"
```

Ignore spaces and uppercase/lowercase differences.


````python
words = ["level", "python", "radar", "hello", "madam"]

for word in words:
    if word == word[::-1]:
        print(f"{word} is palindrome")
    else:
        print(f"{word} is not palindrome")

# Bonus:
text = "Never Odd Or Even"
cleaned = text.replace(" ", "").lower()
if cleaned == cleaned[::-1]:
    print(f"{text} is palindrome")
else:
    print(f"{text} is not palindrome")
````

### 19. FizzBuzz

For every integer from `1` through `30`:

- divisible by both `3` and `5` → `"FizzBuzz"`
- divisible only by `3` → `"Fizz"`
- divisible only by `5` → `"Buzz"`
- otherwise → print the number

Example:

```text
1
2
Fizz
4
Buzz
Fizz
...
14
FizzBuzz
```

The main challenge is figuring out **which `if` condition has to come first**.


````python
for i in range(1, 31):
    if i % 3 == 0 and i % 5 == 0:
        print("FizzBuzz")
    elif i % 3 == 0:
        print("Fizz")
    elif i % 5 == 0:
        print("Buzz")
    else:
        print(i)
````

### 20. Online Exam Result System

A student has answers recorded as:

```python
answers = [True, True, False, True, False, True, True, False, True, True]
```

Each `True` answer is worth **1 point**.  
Each `False` answer is worth **0 points**.

Using a loop, calculate the score.

Then assign a result:

- below 5 → `"Fail"`
- 5–7 → `"Pass"`
- 8–9 → `"Very Good"`
- 10 → `"Perfect"`

Print something like:

```text
Correct answers: 7
Wrong answers: 3
Score: 7
Result: Pass
```

Then calculate:

```python
passed = score >= 5
```

and print:

```text
Passed: True
```

or:

```text
Passed: False
```


````python
answers = [True, True, False, True, False, True, True, False, True, True]

correct = 0
wrong = 0

for answer in answers:
    if answer:
        correct += 1
    else:
        wrong += 1

score = correct

if score < 5:
    result = "Fail"
elif 5 <= score <= 7:
    result = "Pass"
elif 8 <= score <= 9:
    result = "Very Good"
else:
    result = "Perfect"

passed = score >= 5

print(f"Correct answers: {correct}")
print(f"Wrong answers: {wrong}")
print(f"Score: {score}")
print(f"Result: {result}")
print(f"Passed: {passed}")
````
